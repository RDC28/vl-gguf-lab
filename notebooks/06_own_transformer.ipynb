## Configuration

Use the same settings to resume. Saved run folders are printed before inference starts. Set `RESUME` to that folder name after reconnecting; completed outputs are recovered from Drive.

Training checkpoint tradeoff: full checkpoints of all decoder weights plus AdamW state are about 1.6 GB, too large to rewrite on a free 15 GB Drive (replaced files count against quota in Trash). They are saved every 10 steps to Colab local disk only. Drive keeps one fixed-name float16 weight snapshot, `weights-snapshot.pt` (about 270 MB), replaced once per epoch, so all Drive checkpoint writes stay under 1 GB. Resume uses the local full checkpoint when the same runtime still has it, with exact optimizer recovery. After a runtime reset it falls back to the last Drive snapshot: up to one epoch of training is repeated, weights are rounded to float16 and the AdamW optimizer restarts. Every snapshot restore is recorded in `snapshot-restores.jsonl` and `training-summary.json`. The after-training evaluation always uses the final float16 snapshot weights, so every after answer comes from the same saved model, including after a reconnect mid-evaluation.
